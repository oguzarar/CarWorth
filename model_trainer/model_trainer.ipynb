```python
import pandas as pd
from catboost import CatBoostRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
```

```python
cd=pd.read_csv("cars1.csv")
```

```python
cd.head()
```

```python
X = cd[["konum","marka","seri","model","yil","kilometre","vites_tipi","yakit_tipi","kasa_tipi","motor_hacmi","motor_gucu","cekis","ortalama_yakit_tuketimi","yakit_deposu","tramer","degisen","boyali"]]

y = cd["fiyat"]
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=15)
```

```python
kategori=["konum","marka","seri","model","yil","vites_tipi","yakit_tipi","kasa_tipi","cekis"]

model = CatBoostRegressor(cat_features=kategori, verbose=0)
model.fit(X_train, y_train)

model.save_model("araba_fiyat_model.cbm")
```

```python
y_pred = model.predict(X_test)
```

```python
X_test
```

```python
r2_score(y_test,y_pred)
```
